```python
def evaluate_labels(gt, pred):

    gt_set = set(gt)
    pred_set = set(pred)

    common = gt_set.intersection(pred_set)
    correct_predictions = len(common)

    predicted_tokens = len(pred_set)
    input_tokens = len(gt_set)

    accuracy = correct_predictions / input_tokens
    precision = correct_predictions / predicted_tokens
    recall = correct_predictions / input_tokens

    if precision + recall == 0:
        f1 = 0
    else:
        f1 = 2 * (precision * recall) / (precision + recall)

    return accuracy, precision, recall, f1, common

```
